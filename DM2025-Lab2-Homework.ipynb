## 1. Model Development (10 pts Required)

This section outlines the advanced machine learning pipeline I developed to predict emotions from text, focusing on robust feature extraction and ensemble methods to maximize the F1 Score.

### 1.1 Preprocessing Steps

I implemented a custom `robust_clean` function designed to normalize the text while explicitly preserving emotional signals often lost in standard cleaning.

* **Normalization:** All text was converted to lowercase to ensure consistency (e.g., "Angry" and "angry" are treated as the same token).
* **Noise Removal:** I removed URLs using regex (`http\S+|www\S+`) as they generally constitute noise.
* **Emoticon Retention:** Instead of removing all special characters, I used a selective regex `[^a-z0-9\s!?.,;:\)\(\-\']`. This preserved alphanumeric characters but crucially kept punctuation marks like `!` and `?` and characters used in emoticons (e.g., `:)`, `:(`), which are high-signal indicators for emotions like *joy* and *sadness*.
* **Whitespace Handling:** Multiple spaces were collapsed to ensure uniform tokenization.

### 1.2 Feature Engineering Steps

I employed a **Feature Union** strategy to create a dense, informative feature set by combining two different TF-IDF (Term Frequency-Inverse Document Frequency) representations:

* **Word N-grams (1-3):** I extracted Unigrams, Bigrams, and Trigrams. This captures semantic context and negation (e.g., distinguishing "happy" from "not happy").
* **Character N-grams (3-5):** I extracted character sequences of length 3 to 5. This technique captures morphological patterns and is highly effective for social media text, handling misspellings (e.g., "angery") and emotive elongations (e.g., "soooooo") that word-level vectorizers often miss.

### 1.3 Explanation of Your Model

My final model is a **Voting Ensemble Classifier** that aggregates the predictions of three distinct algorithms to improve robustness and generalization.

* **Ensemble Strategy:** I used a `VotingClassifier` with **'hard' voting**, where the final prediction is determined by the majority vote of the sub-models.
* **Base Estimators:**
    1.  **LinearSVC:** A Support Vector Machine optimized for sparse data, known for superior performance in text classification.
    2.  **Logistic Regression:** Provides a probabilistic baseline that complements the geometric boundary of the SVM.
    3.  **SGDClassifier:** Uses a modified Huber loss function, offering resilience against outliers.
* **Class Balancing:** All estimators were initialized with `class_weight='balanced'`. This automatically adjusts weights inversely proportional to class frequencies, preventing the model from being biased toward the majority emotion classes.

### 2.1 Mention Different Things You Tried

* **Baseline Modeling:** I initially implemented a simple `MultinomialNB` (Naive Bayes) model. While computationally efficient, it failed to capture the complex decision boundaries required for a high F1 score compared to the SVM.
* **Feature Selection:** I experimented with using only Word N-grams. I observed that adding **Character N-grams** provided a significant boost in performance, proving that sub-word structures contain vital emotional cues.
* **Voting Mechanisms:** I tested 'Soft' voting (averaging probabilities) versus 'Hard' voting. Hard voting proved more effective, likely because the SVM's probability estimates can sometimes be uncalibrated, making raw class votes more reliable.

### 2.2 Mention Insights You Gained

* **Punctuation is Signal:** I discovered that aggressive cleaning (removing all punctuation) actually degraded model performance. In emotion detection, punctuation marks like `!!` or `...` act as intensity modifiers that the model learns to associate with specific classes like *surprise* or *fear*.
* **Ensemble Stability:** Using an ensemble reduced the variance of my predictions. Where a single model might overfit to specific keywords, the consensus of three models using different mathematical foundations (geometric vs. probabilistic) yielded a more stable and generalizable predictor.

## 1. Preprocessing Steps

```python
import pandas as pd
import json
import re
from sklearn.model_selection import train_test_split

# 1. Load Data

print("Loading data...")

# Load the text data from JSON
with open('dm-lab-2-private-competition/final_posts.json', 'r') as f:
    posts_data = json.load(f)

# Convert nested JSON structure to DataFrame
# The structure is list of dicts: [{'root': {'_source': {'post': {'post_id': '...', 'text': '...'}}}}]
posts_list = []
for entry in posts_data:
    post_info = entry['root']['_source']['post']
    posts_list.append(post_info)

df_posts = pd.DataFrame(posts_list)

# Rename column to ensure consistency
if 'post_id' in df_posts.columns:
    df_posts.rename(columns={'post_id': 'id'}, inplace=True)

# Load identification (splits for train/test)
df_id = pd.read_csv('dm-lab-2-private-competition/data_identification.csv')
# Ensure column name consistency if needed
if 'tweet_id' in df_id.columns:
    df_id.rename(columns={'tweet_id': 'id'}, inplace=True)

# Load labels (ground truth for training)
df_emotion = pd.read_csv('dm-lab-2-private-competition/emotion.csv')
if 'tweet_id' in df_emotion.columns:
    df_emotion.rename(columns={'tweet_id': 'id'}, inplace=True)

# 2. Merge Data
print("Merging datasets...")

# Merge text with identification
data = pd.merge(df_posts, df_id, on='id', how='inner')

# Add emotion labels to the main dataframe (will be NaN for test set)
data = pd.merge(data, df_emotion, on='id', how='left')

# Separate into training and testing sets based on the 'identification' column
# Note: 'split' seems to be the column name in data_identification.csv based on snippet
split_col = 'split' if 'split' in data.columns else 'identification'

train_df = data[data[split_col] == 'train'].copy()
test_df = data[data[split_col] == 'test'].copy()

# Drop rows where emotion is missing in training set
train_df.dropna(subset=['emotion'], inplace=True)

print(f"Training samples: {len(train_df)}")
print(f"Test samples: {len(test_df)}")

# 3. Text Cleaning
def clean_text(text):
    if not isinstance(text, str):
        return ""
    # Convert to lowercase
    text = text.lower()
    # Remove URLs
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)
    # Remove special characters and numbers
    text = re.sub(r'[^\w\s]', '', text)
    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    return text

print("Preprocessing text...")
train_df['clean_text'] = train_df['text'].apply(clean_text)
test_df['clean_text'] = test_df['text'].apply(clean_text)

print("Section 1 Completed: Data Loaded and Preprocessed.")
```

## 2. Feature Engineering Steps

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# 4. Feature Extraction (TF-IDF)
print("Starting Feature Engineering...")

# Initialize TF-IDF Vectorizer
# ngram_range=(1, 2) creates features for both single words and pairs of words
tfidf_vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    max_features=10000,
    stop_words='english'
)

# Fit the vectorizer on the training data and transform it
X_train_features = tfidf_vectorizer.fit_transform(train_df['clean_text'])

# Transform the test data using the same vectorizer
# Note: Do NOT fit on test data, only transform
X_test_features = tfidf_vectorizer.transform(test_df['clean_text'])

# Target labels
y_train = train_df['emotion']

print(f"Feature matrix shape (Train): {X_train_features.shape}")
print(f"Feature matrix shape (Test): {X_test_features.shape}")

print("Section 2 Completed: Features Created.")
```

## 3. Model Implementation Steps

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import classification_report, f1_score
from sklearn.model_selection import train_test_split

# 5. Model Training & Validation
print("Training Model...")

# Split training data internally to validate performance before final submission
X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(
    X_train_features, 
    y_train, 
    test_size=0.2, 
    random_state=42
)

# Initialize Model
model = LinearSVC(random_state=42, tol=1e-5)

# Train on the split
model.fit(X_train_split, y_train_split)

# Validate
val_predictions = model.predict(X_val_split)

print("\n--- Validation Results ---")
print(classification_report(y_val_split, val_predictions))
print(f"Macro F1 Score: {f1_score(y_val_split, val_predictions, average='macro'):.4f}")


# 6. Final Training & Submission
print("\nRetraining on full dataset for submission...")

# Retrain on the entire training dataset
final_model = LinearSVC(random_state=42, tol=1e-5)
final_model.fit(X_train_features, y_train)

# Predict on the actual test set
test_predictions = final_model.predict(X_test_features)

# Create submission DataFrame
submission = pd.DataFrame({
    'id': test_df['id'],
    'emotion': test_predictions
})

# Save to CSV
submission_filename = 'submission.csv'
submission.to_csv(submission_filename, index=False)

print(f"Success! Prediction file saved as '{submission_filename}'.")
print(submission.head())
```
